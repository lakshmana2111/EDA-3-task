```python
df["date"] = pd.to_datetime(df["date"],utc=True)
```

```python
print(df.isnull().sum())
```

```python
df=df.dropna()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
df
```

```python
df=df.sort_values("date")
```

```python
df["Daily_Delta"]=df["close"] - df["open"]
```

```python
df["Daily_return"]= (
    (df["close"] - df["open"])/df["open"]
)*100
```

```python
print(df[
    ["date","open","close","Daily_Delta","Daily_return"]
].head())
```

```python
avg_volume=df["volume"].mean()
max_volume=df["volume"].max()
min_volume=df["volume"].min()
print("Average_volume:", avg_volume)
print("Maximum_volume:", max_volume)
print("Minimum_volume:", min_volume)
```

```python
threshold=2* avg_volume
```

```python
anomalous_days=df[df["volume"]>threshold]
```

```python
print("\nAnamalous Trading Days:")
print(
    anomalous_days.head()[
        [
            "date",
            "open",
            "close",
            "volume",
            "Daily_Delta",
            "Daily_return"
        ]
    ]
)
```

```python
print("Mean Return:", df["Daily_return"].mean())
print("Variance:", df["Daily_return"].var())
print("Standard Deviation:", df["Daily_return"].std())
```
